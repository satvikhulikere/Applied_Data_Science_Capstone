```python
# Import required libraries
import pandas as pd
import folium
from folium.plugins import MarkerCluster
from folium.features import DivIcon
from js import fetch
import io

# -------------------------------
# Load SpaceX GEO dataset
# -------------------------------
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv'
resp = await fetch(URL)
spacex_csv_file = io.BytesIO((await resp.arrayBuffer()).to_py())
spacex_df = pd.read_csv(spacex_csv_file)

# -------------------------------
# Helper function for marker color
# -------------------------------
def get_marker_color(class_value):
    return 'green' if class_value == 1 else 'red'

# -------------------------------
# Create base map
# -------------------------------
nasa_coordinate = [29.559684888503615, -95.0830971930759]
site_map = folium.Map(location=nasa_coordinate, zoom_start=5)

# -------------------------------
# Create marker cluster
# -------------------------------
marker_cluster = MarkerCluster()

# -------------------------------
# Add markers for each launch
# -------------------------------
for _, row in spacex_df.iterrows():
    if pd.notnull(row['Lat']) and pd.notnull(row['Long']):
        folium.CircleMarker(
            location=[row['Lat'], row['Long']],
            radius=6,
            color=get_marker_color(row['class']),
            fill=True,
            fill_color=get_marker_color(row['class']),
            fill_opacity=0.7,
            popup=(
                f"Launch Site: {row['Launch Site']}<br>"
                f"Launch Outcome: {'Success' if row['class'] == 1 else 'Failure'}"
            )
        ).add_to(marker_cluster)

# -------------------------------
# Add cluster to map
# -------------------------------
site_map.add_child(marker_cluster)

# Display map
site_map

```

```python
# TASK 3: Calculate the distances between a launch site to its proximities
from math import sin, cos, sqrt, atan2, radians
def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

launch_site = launch_sites_df.loc[
    launch_sites_df['Launch Site'] == 'CCAFS LC-40'
]

launch_site_lat = launch_site['Lat'].values[0]
launch_site_long = launch_site['Long'].values[0]

coastline_lat = 28.56367
coastline_long = -80.57163

distance_coastline = calculate_distance(
    launch_site_lat,
    launch_site_long,
    coastline_lat,
    coastline_long
)

distance_coastline

# Create a map centered at the launch site
distance_map = folium.Map(
    location=[launch_site_lat, launch_site_long],
    zoom_start=13
)

# Draw a line between launch site and coastline
folium.PolyLine(
    locations=[
        [launch_site_lat, launch_site_long],
        [coastline_lat, coastline_long]
    ],
    color='blue'
).add_to(distance_map)

# Add distance label
folium.Marker(
    location=[coastline_lat, coastline_long],
    icon=DivIcon(
        icon_size=(200,36),
        icon_anchor=(0,0),
        html=f'<div style="font-size: 12px; color: blue;">{distance_coastline:.2f} km</div>'
    )
).add_to(distance_map)

distance_map



```

```python
from math import sin, cos, sqrt, atan2, radians
import folium
from folium.features import DivIcon
from folium.plugins import MousePosition

# 1️⃣ Define a function to calculate distance between two coordinates
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0  # approximate radius of Earth in km
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c

# 2️⃣ Base map centered at launch site (example: Kennedy Space Center)
launch_site_lat = 28.573255
launch_site_lon = -80.646895
site_map = folium.Map(location=[launch_site_lat, launch_site_lon], zoom_start=10)

# Add MousePosition to explore coordinates
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)
site_map.add_child(mouse_position)

# 3️⃣ Define coordinates for points of interest (replace with your MousePosition values)
# Example coordinates
coastline_lat, coastline_lon = 28.56367, -80.57163
city_lat, city_lon = 28.485, -80.526
railway_lat, railway_lon = 28.580, -80.600
highway_lat, highway_lon = 28.570, -80.620

poi_list = [
    ("Coastline", coastline_lat, coastline_lon),
    ("City", city_lat, city_lon),
    ("Railway", railway_lat, railway_lon),
    ("Highway", highway_lat, highway_lon)
]

# 4️⃣ Add markers and draw lines to launch site
for poi_name, lat, lon in poi_list:
    # Calculate distance
    distance_km = calculate_distance(launch_site_lat, launch_site_lon, lat, lon)
    
    # Add a marker at the POI with distance displayed
    folium.Marker(
        location=[lat, lon],
        icon=DivIcon(
            icon_size=(150,36),
            icon_anchor=(0,0),
            html=f'<div style="font-size: 12px; color:#d35400;"><b>{poi_name}: {distance_km:.2f} KM</b></div>',
        )
    ).add_to(site_map)
    
    # Draw a line between launch site and the POI
    line = folium.PolyLine(
        locations=[[launch_site_lat, launch_site_lon], [lat, lon]],
        weight=2,
        color='blue',
        opacity=0.6
    )
    site_map.add_child(line)

# 5️⃣ Add a marker for the launch site itself
folium.Marker(
    location=[launch_site_lat, launch_site_lon],
    icon=DivIcon(
        icon_size=(150,36),
        icon_anchor=(0,0),
        html='<div style="font-size: 12px; color:red;"><b>Launch Site</b></div>',
    )
).add_to(site_map)

# Display the map
site_map

```
